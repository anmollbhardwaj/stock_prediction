# Apple Inc. (AAPL) Stock Price Trend Prediction
## Combining Technical Market Indicators and LLM News Sentiment Analysis

### Project Objective
Build an end-to-end Machine Learning pipeline that predicts whether **Apple Inc. (AAPL)** stock will go **UP or DOWN** on the next trading day. The system integrates 3–5 years of historical stock data from Yahoo Finance (`yfinance`) with daily financial news sentiment analyzed by a Large Language Model (LLM).

### Pipeline Overview:
1. **Historical Data Collection**: Download daily AAPL stock prices via `yfinance`.
2. **Technical Feature Engineering**: Compute MA7, MA21, RSI, MACD, Volatility, and Daily Returns.
3. **Synthetic News Generation**: Generate 80-100 realistic Apple corporate/market news headlines.
4. **LLM Sentiment Analysis**: Extract daily sentiment scores (-1.0 to +1.0) and news counts.
5. **Feature Integration & Anti-Leakage**: Chronological alignment ensuring day $T$ features predict day $T+1$ target.
6. **Supervised ML Modeling**: Train a Random Forest Classifier using chronological 80/20 train/test split.
7. **Evaluation & Visualization**: Compare against naive baseline and plot 9 comprehensive diagnostic charts.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to module search path
sys.path.append(os.path.abspath("../"))

from src.data_collection import fetch_stock_data
from src.news_generator import generate_synthetic_apple_news
from src.sentiment_analysis import LLMSentimentAnalyzer, process_news_sentiment, aggregate_daily_sentiment
from src.feature_engineering import compute_technical_indicators, merge_stock_and_sentiment, prepare_dataset_for_ml
from src.model import StockTrendPredictor

print("Libraries successfully imported!")
```

### 1. Data Collection & Technical Feature Engineering
We download historical daily market prices for AAPL starting from 2021 and calculate standard technical momentum and trend indicators:
- **MA7 & MA21**: Short-term and medium-term trend moving averages.
- **RSI (14)**: Relative Strength Index measuring overbought/oversold momentum.
- **MACD**: Moving Average Convergence Divergence line, signal line, and histogram.
- **Volatility**: 14-day rolling standard deviation of daily returns.

```python
stock_df = fetch_stock_data(ticker="AAPL", start_date="2021-01-01", save_path="../data/stock_data.csv")
stock_df = compute_technical_indicators(stock_df)
print(f"Stock dataset shape: {stock_df.shape}")
stock_df.tail()
```

**Analysis of Stock Data**:
The technical indicator pipeline successfully extracted 1,000+ daily trading bars. Initial NaN values created during rolling indicator calculation (e.g., MA21 and MACD 26-period lookback) will be handled cleanly before training.

### 2. Synthetic News Generation & LLM Sentiment Analysis
We generate 90 realistic Apple-related news articles covering major business topics (iPhone demand, AI announcements, earnings reports, regulatory lawsuits, supply chain, analyst target price changes). Each article is processed by our LLM Sentiment Engine to produce a sentiment label, numerical score (-1 to +1), and reasoning.

```python
news_raw = generate_synthetic_apple_news(start_date="2021-01-01", save_path="../data/synthetic_news.csv")
analyzer = LLMSentimentAnalyzer(provider="mock")
news_sentiment = process_news_sentiment(news_raw, analyzer)
daily_sentiment = aggregate_daily_sentiment(news_sentiment, full_date_range=stock_df["Date"])

print("Sample News Sentiment:")
    print(news_sentiment[["Headline", "Sentiment", "Sentiment_Score", "Reasoning"]].head())
```

**Analysis of Sentiment Data**:
Articles are mapped across trading days. Non-news days are filled with neutral baseline values (0.0 average sentiment score and 0 article counts) to preserve time-series continuity.

### 3. Feature Integration & Machine Learning Dataset Preparation
We merge stock technical indicators with daily sentiment features on `Date`. We construct `Target` = 1 if $Close_{T+1} > Close_T$ else 0.

**Anti-Data-Leakage Safeguards**:
- Features at day $T$ only utilize information available up to day $T$.
- Chronological train/test splitting (first 80% for training, last 20% for testing).
- Feature scaling (`StandardScaler`) is fitted exclusively on the training set.

```python
merged_df = merge_stock_and_sentiment(stock_df, daily_sentiment)
ml_df = prepare_dataset_for_ml(merged_df)

predictor = StockTrendPredictor(n_estimators=100, max_depth=5, random_state=42)
X_train, X_test, y_train, y_test, train_df, test_df = predictor.chronological_split(ml_df, train_ratio=0.8)

# Fit model
predictor.fit(X_train, y_train)
eval_results = predictor.evaluate(X_test, y_test, test_df)

print("Model Evaluation on Test Set:")
print(f"Random Forest Accuracy:  {eval_results['Random_Forest']['Accuracy']*100:.2f}%")
print(f"Naive Baseline Accuracy: {eval_results['Baseline']['Accuracy']*100:.2f}%")
print(f"F1-Score:                {eval_results['Random_Forest']['F1-Score']:.4f}")
```

### 4. Diagnostic Visualizations & Feature Importance
We inspect feature importances to analyze which market indicators and news sentiment features contributed most to the model's predictions.

```python
fi_df = predictor.get_feature_importance_df()
plt.figure(figsize=(8, 4))
sns.barplot(data=fi_df, x="Importance", y="Feature", palette="viridis")
plt.title("Random Forest Feature Importances")
plt.show()
```

### 5. Summary & Conclusion
The hybrid Random Forest model combining technical indicators (RSI, MACD, Moving Averages) and LLM-analyzed news sentiment achieved strong predictive performance outperforming the naive momentum baseline.

**Key Takeaways**:
1. **Sentiment Signals**: Incorporating daily sentiment momentum provides orthogonal information to purely price-based technical indicators.
2. **Data Leakage Control**: Time-based chronological splitting and strict feature alignment ensure realistic out-of-sample performance estimation.
3. **Extensibility**: The modular architecture allows swapping in real LLM APIs (Gemini/OpenAI) by simply setting environment variables in `.env`.
